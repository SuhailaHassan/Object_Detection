# Detect objects in video

# Detect People/Pedestrians

```python
import cv2
import numpy as np

# Video with people in it
cap = cv2.VideoCapture('Pedestrians.mp4')
# Create our pedesterian classifier & Enter XML file
pedesterian = cv2.CascadeClassifier('haarcascade_fullbody.xml')

# While video is playing
while cap.isOpened():
    # Read each frame of the video 
    ret, frame = cap.read()
    # Convert frame to grayscale
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    # Pass frame in our pedesterian classifier to detectMultiScale  
    pedesterians = pedesterian.detectMultiScale(frame,1.2,3)
    # Put the pedesterians in a rectangular box to identify them in video
    for (x,y,w,h) in pedesterians:
        cv2.rectangle(frame,(x, y),(x+w, y+h),(0,255,0),2)
        cv2.imshow('Pedestrians', frame)
    if cv2.waitKey(1) == 5:
        break

cap.release()
cv2.destroyAllWindows()
```

# Detect Cars

```python
import cv2
import numpy as np
import time

# Video with cars in it
cap = cv2.VideoCapture('Cars.mp4')
# Create our car classifier & Enter XML file
car = cv2.CascadeClassifier('haarcascade_car.xml')

# While video is playing
while cap.isOpened():
    time.sleep(.05)
    # Read each frame of the video 
    ret, frame = cap.read()
    # Convert frame to grayscale
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    # Pass frame in our car classifier to detectMultiScale  
    cars = car.detectMultiScale(frame,2)
    # Put the cars in a rectangular box to identify them in video
    for (x,y,w,h) in cars:
        cv2.rectangle(frame,(x, y),(x+w, y+h),(0, 255, 0),2)
        cv2.imshow('Cars', frame)
    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

# Detect Peoples' Faces

```python
import cv2
import numpy as np

# Video with faces in it
cap = cv2.VideoCapture('Faces.mp4')
# Create our face classifier & Enter XML file
face = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# While video is playing
while cap.isOpened():
    # Read each frame of the video 
    ret, frame = cap.read()
    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # Pass frame in our face classifier to detectMultiScale  
    faces = face.detectMultiScale(frame,1.5,2)
    # Put the faces in a rectangular box to identify them in video
    for (x,y,w,h) in faces:
        cv2.rectangle(frame,(x, y),(x+w, y+h),(0, 255, 0),2)
        cv2.imshow('Faces',frame)
    if cv2.waitKey(1) == 5:
        break

cap.release()
cv2.destroyAllWindows()
```
